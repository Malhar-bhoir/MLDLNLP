```python
# binary search alogrith 
# function
def binary(arr,target): 
    # size
    size  = len(arr)

    # start ,end
    start = 0 
    end = size - 1

    # while loop
    while(start <= end):
        mid = (start + end)//2

        if(arr[mid] == target): 
            return mid

        elif (arr[mid] < target):
            start = mid + 1 

        elif(arr[mid] > target):
            end = mid -1 
    

    return -1 # if not found 


myy_list = [10,20,30,45,50,60,70,80]

target = 70

result = binary(myy_list,target)

print(result) 
```
