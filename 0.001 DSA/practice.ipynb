5. Roman to Integer

The problem: Convert a Roman numeral string (like “MCMXC”) into its integer value.

Key rule to understand first: Normally you add values left to right (like “VI” = 5+1 = 6). But if a smaller value comes before a bigger one, you subtract instead (like “IV” = 5-1 = 4).


````python
def roman_to_int(s):
    values = {
        'X': 10,
        'L': 50, 
        'C': 100,
        'D': 500,
        'M': 1000
    }

    total = 0 

    for i in range(len(s)):
        if i + 1 < len(s) and values[s[i]] < values[s[i+1]]:
            total = total - values[s[i]]

        else :
        
            total = total + values[s[i]]

    return total

s = "MCMXC"
print(roman_to_int(s))



````

# **2. First, understand how carry works**

Let's understand normal addition.

Suppose:

```
  129
+   1
-----
  130
```

We start adding from the rightmost digit.

- 9 + 1 = 10 → write 0 and carry 1.
- 2 + 1 = 3 → write 3.
- 1 stays the same.

Result = 130.

Now consider:

```
  199
+   1
-----
  200
```

- 9 + 1 = 10 → write 0, carry 1.
- 9 + 1 = 10 → write 0, carry 1.
- 1 + 1 = 2.

Result = 200.

**Our Python program follows exactly this logic.**

# **3. Understand the approach**

We have an array:

```
digits = [1, 2, 9]
```

We start from the last index and move towards the first index.

Why?

Because when we add 1 to a number, the carry starts from the rightmost digit.

Our logic is:

1. Check the last digit.
2. If it is less than 9, add 1 and finish.
3. If it is 9, change it to 0 because the carry moves to the left.
4. Repeat with the previous digit.
5. If every digit was 9, add 1 at the beginning.

# **10. Valid Parentheses**

## **1. Understand the problem first**

We are given a string containing different types of brackets:

- `()` – Round brackets
- – Square brackets
- `{}` – Curly brackets

Our task is to check whether these brackets are properly opened and closed.

Let's understand with examples.

**Example 1:**

```
s = "()"
```

Here:

- `(` opens a bracket.
- `)` closes that bracket.

Valid

**Example 2:**

```
s = "()[]{}"
```

All brackets are properly opened and closed.

Valid

**Example 3:**

```
s = "(]"
```

Here:

- `(` is opened.
- `]` is used to close it.

But `]` cannot close `(`.

Invalid

**Example 4:**

```
s = "([)]"
```

This one is interesting!

We open `(`, then open `[`. But then we try to close `)` before closing `[`.

That is incorrect because the most recently opened bracket must be closed first.

Invalid

So, the main rule is:

**Every opening bracket must have its matching closing bracket, and the brackets must close in the correct order.**

### **How does Stack help us solve Valid Parentheses?**

Whenever we encounter an opening bracket, we push it into the stack.

Whenever we encounter a closing bracket, we check whether it matches the most recently opened bracket.

For example:

2. See `[` → Push into stack.
3. See `]` → Matches `[`, so pop it.
4. See `)` → Matches `(`, so pop it.

At the end, the stack is empty.

That means all brackets were correctly matched!

21. Squares of a Sorted Array

The problem: Given a sorted array that may contain negatives, return the squares of each number, also sorted. [-4, -1, 0, 3, 10] → [0, 1, 9, 16, 100].

The catch: Squaring turns big negatives into big positives (-4 becomes 16), so the result is no longer sorted. The simple fix is to square everything and sort, which is O(n log n).

Smarter idea (two pointers): The largest square is always at one of the two ends of the array, because the biggest negative or the biggest positive sits there. Compare the two ends, put the larger square at the back of the result, and move inward.

````python
def square(arr):
    result = []

    for i in arr:  # i enter every value in  list and squares  it 
        result.append(i ** 2)

        result.sort()

    return result
square([-4, -1, 0, 3, 10])
````
