```python
# linear search code 
# arr == mylsit
def linear(arr,target):
    size = len(arr)
    ## for loop 
    for index in range(0,size):
        if arr[index] == target:
            return index
        

    # if not found retrnn this 
    return -1


mylist = [11,12,13,14,15]
target = 140

result = linear(mylist,target)
print(result)
```
