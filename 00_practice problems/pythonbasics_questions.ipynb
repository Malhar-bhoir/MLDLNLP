```python
def climb_stairs(n):
    if n <= 2:
        return n
    prev2, prev1 = 1, 2          # ways(1)=1, ways(2)=2
    for _ in range(3, n + 1):
        prev2, prev1 = prev1, prev1 + prev2
    return prev1
climb_stairs(5)
```

```python
def hcf_gcf(x,y):
    if (x>y):
        smaller = x 
    else:
        smaller = y 

    for i in range(1 , smaller + 1):
        if (x % i == 0) and (y % i ==0):
            hcf = i 
    return hcf


hcf_gcf(12,30)
    
```
