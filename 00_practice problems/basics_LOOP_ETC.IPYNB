Excellent question! This confuses almost every beginner. The reason is that **Python has two different ways of using functions**.

Let's understand it step by step.

---

# 1. Method Functions (Called on an Object)

Some functions belong to a particular object (string, list, dictionary, etc.).

Example:

```python
s = "Hello World"

words = s.split()
```

Here:

* `s` is a **string object**
* `.split()` is a **method of the string**

Think of it like this:

```
String
   │
   ├── split()
   ├── upper()
   ├── lower()
   ├── replace()
```

So you call the method using:

```python
object.method()
```

Examples:

```python
name = "malhar"

print(name.upper())
print(name.lower())
print(name.replace("a", "@"))
```

Output:

```
MALHAR
malhar
m@lh@r
```

---

# 2. Built-in Functions

Some functions do **not belong to any object**.

They are available everywhere in Python.

Example:

```python
numbers = [1, 2, 3]

print(len(numbers))
```

Here:

* `len()` is a built-in function.
* It works on many different data types.

Similarly:

```python
set()
list()
tuple()
dict()
int()
float()
str()
```

These are **built-in functions (and constructors)**.

Example:

```python
a = set()
b = list()
c = dict()
```

They create new objects.

---

# Why isn't it written like this?

You might wonder:

```python
set.split()
```

This doesn't make sense because `set` is **not a string**.

A set doesn't have a `split()` method.

Similarly:

```python
s = "hello"

s.set()
```

This also doesn't make sense because strings don't have a `set()` method.

---

# Think of it like a person

Imagine a person.

A person has abilities:

```
Person
   │
   ├── walk()
   ├── eat()
   ├── sleep()
```

You would say:

```
person.walk()
person.eat()
```

You **wouldn't** say:

```
walk(person)
```

because walking is something a person does.

---

Now imagine a factory.

A factory creates people.

```
create_person()
```

That's similar to:

```python
set()
list()
dict()
```

They **create** objects.

# A simple rule to remember

Ask yourself:

> **"Is this action something the object itself can do?"**

If **yes**, it is usually a **method**:

```python
s.split()
s.upper()
l.append()
d.keys()
```

If **no**, and it is something Python provides generally (like creating an object or calculating something), it is usually a **built-in function**:

```python
len(s)
set()
list()
dict()
max(numbers)
min(numbers)
sum(numbers)
```

### Quick comparison

| Syntax             | Type                            | Example                                        |
| ------------------ | ------------------------------- | ---------------------------------------------- |
| `object.method()`  | Method (belongs to the object)  | `"hello".split()`, `my_list.append(5)`         |
| `function(object)` | Built-in function               | `len(my_list)`, `max(numbers)`, `sum(numbers)` |
| `type()`           | Constructor (creates an object) | `set()`, `list()`, `dict()`, `str()`           |

This distinction—**methods belong to objects, while built-in functions are called independently**—is one of the most important concepts in Python and will help you understand the language much more naturally.
