```python
# finding missing number using xor method 
# " ^ " this symbol indicate the  XOR operation
# to do with XOR we have to make use of forr loop 

def missing_xor(a):
    n = len(a)
    xor_a = a[0] # 1 
    for i in range(1 ,n):
        xor_a = xor_a ^ a[i] # 1 ^ 1 = 0 , 0 ^ 2 = 2 ...


    x2 = 0
    for i in range(1,n+2): # n +2 = 7 
        x2 = x2 ^ i # 0 ^ 1 = 1 , 0 ^ 2 = 2

    print(xor_a ^ x2)
        
    


a = [1,2,4,5,6,7]
missing_xor(a)
    
```
